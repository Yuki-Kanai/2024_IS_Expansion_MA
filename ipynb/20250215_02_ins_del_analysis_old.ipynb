## Hotspot analysis
Find hotspots of insertions by sliding window.

### Sliding window analysis

````julia
# function that inputs the insertion positions and outputs the rate of insertions in each bin
R"""
sliding_window_analysis <- function(pos, bin_size = 100000, step_size =1000, genome_size = ref_genome_size, alpha = 0.05, binom_test = TRUE){
  mutations <- data.frame(position = pos) #data.frame(position = tmp.df %>% pull(pos))

  # Define genome start and end (adjust based on your data)
  genome_start <- 1
  genome_end <- ref_genome_size

  # Handle mutations near the boundaries of the circular genome
  # By adding/subtracting ref_genome_size for wrap-around 
  mutations_extended <- rbind(mutations,
                              data.frame(position = mutations$position + genome_end),
                              data.frame(position = mutations$position - genome_end))

  # Sliding window analysis on extended positions
  windows_ <- data.frame(start = seq(genome_start, ref_genome_size, by = step_size))
  windows_$end <- windows_$start + bin_size - 1

  windows_$mutation_count <- sapply(1:nrow(windows_), function(i) {
    sum(mutations_extended$position >= windows_$start[i] & mutations_extended$position <= windows_$end[i])
  })

  # Calculate mutations per base pair
  windows_$mutations_per_bp <- windows_$mutation_count / bin_size

  p_ <- ggplot(windows_, aes(x=start, y=mutations_per_bp)) +
    geom_line(aes(x=end, y=mutations_per_bp)) + # Use both start and end for a clearer visualization
    #geom_point() +
    xlab('Genomic Position') +
    ylab('Mutations per Base Pair') +
    theme_minimal()
  show(p_)

  mutation_rate <- nrow(mutations) / (ref_genome_size - genome_start + 1)

  if (binom_test){
  windows_$p_value <- sapply(1:nrow(windows_), function(i) {
    binom.test(windows_$mutation_count[i], bin_size, mutation_rate, alternative = "two.sided")$p.value
  })
  windows_$adjusted_p_value <- p.adjust(windows_$p_value, method = "BH")
  windows_$significant <- windows_$adjusted_p_value < alpha
  windows_$test_side <- ifelse(windows_$mutation_count > bin_size * mutation_rate, "high", "low")
  windows_ %>% filter(significant) %>% arrange(adjusted_p_value) %>% head %>% print
  }
  return(list(windows_, mutation_rate))
}
""";
````
